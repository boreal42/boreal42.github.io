## 6. Le GPU fait-il une différence?

Mesure le temps d'**une** époque avec des lots de 256, sur le GPU puis sur le processeur (`tf.device("/CPU:0")`). Pour un si petit modèle, la différence peut te surprendre : le GPU n'est avantageux que quand il y a beaucoup de calcul à paralléliser. Tu le reverras au cours 30, avec les réseaux convolutionnels.

```python
import time

def une_epoque():
    m = keras.Sequential([keras.Input(shape=(784,)), keras.layers.Dense(10, activation="softmax")])
    m.compile(optimizer=keras.optimizers.SGD(learning_rate=0.5), loss="sparse_categorical_crossentropy")
    debut = time.time()
    m.fit(x_train_plat, y_train, epochs=1, batch_size=256, verbose=0)
    return time.time() - debut

# À toi : le temps sur l'appareil par défaut (le GPU, s'il y en a un), puis dans un bloc with tf.device("/CPU:0")
```
