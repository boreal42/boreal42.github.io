```python
import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "shap", "lime"], check=False)
import numpy as np
import matplotlib.pyplot as plt
import shap
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import HistGradientBoostingRegressor
maisons = fetch_california_housing(as_frame=True)
X, y = maisons.data, maisons.target
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0)
modele = HistGradientBoostingRegressor(random_state=0).fit(X_tr, y_tr)
print("R² au test :", round(modele.score(X_te, y_te), 3))
print(list(X.columns))
```
