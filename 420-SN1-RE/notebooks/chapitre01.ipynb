## 2. Python comme calculatrice scientifique

| Opération | Symbole | Exemple | Résultat |
|---|---|---|---|
| Addition | `+` | `7 + 3` | `10` |
| Soustraction | `-` | `7 - 3` | `4` |
| Multiplication | `*` | `7 * 3` | `21` |
| Division | `/` | `7 / 2` | `3.5` |
| Puissance | `**` | `2 ** 10` | `1024` |
| Division entière | `//` | `7 // 2` | `3` |
| Reste (modulo) | `%` | `7 % 2` | `1` |

Les priorités sont celles des mathématiques : parenthèses, puissances, multiplications et divisions, puis additions et soustractions.

**Attention :** Python utilise le **point** comme séparateur décimal. On écrit `9.81`, jamais `9,81`.

```python
# Accélération gravitationnelle (m/s²) et temps de chute (s)
print(9.81 * 2.0 ** 2 / 2)   # distance parcourue en chute libre : g·t²/2
```

## 4. La démarche : planifier, coder, vérifier

C'est la méthode du cours, du premier labo jusqu'à l'épreuve finale. **On ne code jamais avant d'avoir planifié.**

### Planifier
Remplis la fiche de planification :

| Question | Réponse pour °C → °F |
|---|---|
| **Intrants** : quelles données le programme reçoit-il? | une température en °C |
| **Extrants** : que doit-il produire? | la même température en °F |
| **Traitements** : quel calcul relie les deux? | F = C × 9/5 + 32 |
| **Jeu d'essais** : quels résultats attends-tu, calculés à la main? | voir ci-dessous |

Le **jeu d'essais** se construit *avant* de coder, avec des cas que tu sais calculer et quelques cas limites :

| Intrant (°C) | Extrant attendu (°F) | Pourquoi ce cas |
|---|---|---|
| 0 | 32 | point de congélation de l'eau |
| 100 | 212 | point d'ébullition de l'eau |
| −40 | −40 | les deux échelles se croisent |
| 37 | 98.6 | température du corps humain |

### Coder

```python
celsius = 37
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)
```

### Vérifier
Change la valeur de `celsius` pour chaque ligne du jeu d'essais et compare le résultat à ta réponse attendue. Si une seule ligne ne concorde pas, le programme est faux, même si les autres passent.

> Essaie `celsius = 36.6`. Tu obtiens `97.88000000000001` au lieu de `97.88`. Ce n'est pas une erreur de ta part : l'ordinateur représente les nombres décimaux en binaire, avec une précision limitée. On apprendra à arrondir au chapitre 2.

## 6. Les erreurs sont normales

Python s'arrête et affiche un message quand il ne comprend pas une instruction. **Lis le message** : la dernière ligne dit quel type d'erreur, et une flèche montre souvent où.

Exécute la cellule suivante. Elle contient une **erreur de syntaxe** volontaire. Corrige-la.

```python
print("Température :" 25)
```

| Message | Signification habituelle |
|---|---|
| `SyntaxError` | Python ne peut pas lire l'instruction : virgule, parenthèse ou guillemet manquant |
| `NameError` | un nom est utilisé avant d'avoir été créé, ou mal écrit |
| `ZeroDivisionError` | une division par zéro |

### Piège propre à Colab : l'ordre d'exécution
Colab exécute les cellules **dans l'ordre où tu cliques**, pas dans l'ordre de la page. Si tu obtiens un `NameError` sur une variable qui existe plus haut, exécute la cellule qui la crée. En cas de doute : **Exécution → Tout exécuter**.
