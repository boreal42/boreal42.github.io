### Comment fonctionnent les vérifications
Chaque exercice a une cellule de départ, où tu remplaces les `None` par ta réponse, puis une cellule de **vérification** qui te dit ✅ ou ❌. Elle ne montre pas la réponse attendue.

**Exécute d'abord la cellule suivante**, une seule fois : elle prépare l'outil de vérification. Si tu rouvres le cahier plus tard, exécute-la de nouveau.

````python
# Outil de vérification — exécute cette cellule une fois, au début. Ne pas modifier.
import hashlib, math

def _empreinte(v):
    if isinstance(v, bool) or v is None:
        texte = repr(v)
    elif isinstance(v, (int, float)):
        texte = f"{round(float(v), 4):.4f}"
        if texte == "-0.0000":
            texte = "0.0000"
    else:
        texte = repr(v)
    return hashlib.sha256(texte.encode()).hexdigest()[:12]

def verifier(obtenu, empreinte, nom="réponse"):
    if obtenu is None:
        print(f"⏳ {nom} : pas encore de réponse.")
    elif _empreinte(obtenu) == empreinte:
        print(f"✅ {nom} : correct ({obtenu!r})")
    else:
        print(f"❌ {nom} : pas encore. Tu as obtenu {obtenu!r}. Vérifie ton calcul, puis réessaie.")
````
