## Problème d'ouverture

Une balle de 0,5 kg roule à 3 m/s au sommet d'une rampe de 2 m de haut. Quelle est son énergie mécanique?

Tu sais faire ce calcul à la main. L'objectif de ce chapitre : écrire un programme qui le fait pour **n'importe quelle** masse, vitesse et hauteur, et qui présente le résultat proprement.

**Exécute d'abord la cellule suivante.** Elle prépare l'outil qui vérifie tes réponses aux exercices.

````python
# Outil de vérification — exécute cette cellule une fois, au début. Ne pas modifier.
import hashlib, math

def _empreinte(v):
    if isinstance(v, bool) or v is None:
        texte = repr(v)
    elif isinstance(v, (int, float)):
        texte = f"{round(float(v), 4):.4f}"
        if texte == "-0.0000":
            texte = "0.0000"
    else:
        texte = repr(v)
    return hashlib.sha256(texte.encode()).hexdigest()[:12]

def verifier(obtenu, empreinte, nom="réponse"):
    if obtenu is None:
        print(f"⏳ {nom} : pas encore de réponse.")
    elif _empreinte(obtenu) == empreinte:
        print(f"✅ {nom} : correct ({obtenu!r})")
    else:
        print(f"❌ {nom} : pas encore. Tu as obtenu {obtenu!r}. Vérifie ton calcul, puis réessaie.")
````

## 1. Les variables

Une variable est un **nom** associé à une **valeur**. Le signe `=` est une **affectation** : la valeur à droite est calculée, puis rangée dans le nom à gauche.

````python
masse = 0.5        # kg
vitesse = 3.0      # m/s
print(masse, vitesse)
````

Comme `=` veut dire « reçoit », l'instruction suivante est valide, même si elle est fausse en mathématiques : on calcule `compteur + 1`, puis on range le résultat dans `compteur`.

````python
compteur = 10
compteur = compteur + 1
print(compteur)
````

### Bien nommer ses variables

| Règle | Exemple valide | Exemple invalide |
|---|---|---|
| Lettres, chiffres et `_` seulement | `vitesse_initiale` | `vitesse-initiale` |
| Ne commence pas par un chiffre | `t2` | `2t` |
| Les majuscules comptent | `masse` et `Masse` sont deux variables différentes | |

Dans ce cours, on écrit les noms en minuscules, avec des `_` entre les mots, et on indique l'**unité en commentaire**. En sciences, une valeur sans unité ne veut rien dire.

## 2. Les types de données

Chaque valeur a un **type**. La fonction `type()` le révèle.

| Type | Ce qu'il représente | Exemples |
|---|---|---|
| `int` | nombre entier | `3`, `-12`, `0` |
| `float` | nombre à virgule (réel) | `9.81`, `-0.5`, `6.022e23` |
| `str` | texte (chaîne de caractères) | `"H2O"`, `'pH'` |
| `bool` | vrai ou faux | `True`, `False` |

````python
print(type(42))
print(type(9.81))
print(type("9.81"))
print(type(3 > 2))
````

Remarque deux choses :

- `"9.81"` est du **texte**, pas un nombre. Les guillemets font toute la différence.
- `6.022e23` est la notation scientifique de Python : 6,022 × 10²³.

### Conversions

| Fonction | Convertit en | Exemple | Résultat |
|---|---|---|---|
| `int()` | entier (tronque, n'arrondit pas) | `int(7.9)` | `7` |
| `float()` | réel | `float("9.81")` | `9.81` |
| `str()` | texte | `str(25)` | `"25"` |

````python
print(int(7.9))
print(float("9.81") * 2)
print("Température : " + str(25) + " °C")
````

## 3. Les expressions

Une **expression** combine des valeurs, des variables et des opérateurs. Python la calcule en suivant les priorités mathématiques.

Deux opérateurs servent souvent en sciences pour découper des quantités :

- `//` donne le **quotient entier** : `125 // 60` vaut `2`;
- `%` donne le **reste** : `125 % 60` vaut `5`.

Ensemble, ils convertissent 125 secondes en 2 min 5 s.

````python
duree = 125          # s
minutes = duree // 60
secondes = duree % 60
print(minutes, "min", secondes, "s")
````

Deux fonctions intégrées très utiles :

- `round(x, n)` arrondit `x` à `n` décimales;
- `abs(x)` donne la valeur absolue.

### Piège : la division `/` donne toujours un `float`
`10 / 2` vaut `5.0`, pas `5`. Ce n'est pas une erreur, c'est le comportement normal de Python.

## 4. Le module `math`

Python n'a pas de racine carrée ni de sinus « de base ». Ils sont dans le **module** `math`, qu'il faut **importer** une fois.

````python
import math

print(math.sqrt(16))        # racine carrée
print(math.pi)              # π
print(math.exp(1))          # e¹
print(math.log(math.e))     # logarithme NATUREL (ln)
print(math.log10(1000))     # logarithme en base 10
````

### Piège majeur : les angles sont en **radians**
`math.sin(30)` ne donne **pas** 0,5 : Python interprète 30 comme 30 radians. Il faut convertir avec `math.radians()`.

````python
print(math.sin(30))                  # faux pour 30°
print(math.sin(math.radians(30)))    # correct : 0.5 (ou presque)
````

Tu obtiens `0.49999999999999994` au lieu de `0.5`. C'est la précision limitée des nombres réels. Arrondis à l'affichage, jamais en cours de calcul.

### Les f-strings
Une **f-string** commence par `f` et insère des valeurs entre accolades. Après `:`, on précise le format.

````python
g = 9.81
masse = 2.0
poids = masse * g

print(f"Poids : {poids} N")
print(f"Poids : {poids:.2f} N")        # 2 décimales
print(f"Poids : {poids:.3e} N")        # notation scientifique
print(f"Masse de la Terre : {5.972e24:.2e} kg")
````

## 6. Exemple complet : l'énergie mécanique

### Planifier

| Question | Réponse |
|---|---|
| Intrants | masse *m* (kg), vitesse *v* (m/s), hauteur *h* (m) |
| Extrants | énergies cinétique, potentielle et mécanique (J) |
| Traitements | Ec = ½ m v², Ep = m g h, Em = Ec + Ep, avec g = 9,81 m/s² |

**Jeu d'essais**, calculé à la main avant de coder :

| *m* (kg) | *v* (m/s) | *h* (m) | Em attendue (J) | Pourquoi ce cas |
|---|---|---|---|---|
| 0,5 | 3 | 2 | 12,06 | le problème d'ouverture |
| 1 | 0 | 0 | 0 | objet immobile au sol |
| 2 | 1 | 0 | 1 | énergie cinétique seule |
| 1 | 0 | 1 | 9,81 | énergie potentielle seule |

### Coder

````python
G = 9.81   # m/s², accélération gravitationnelle

masse = 0.5      # kg
vitesse = 3.0    # m/s
hauteur = 2.0    # m

energie_cinetique = 0.5 * masse * vitesse ** 2
energie_potentielle = masse * G * hauteur
energie_mecanique = energie_cinetique + energie_potentielle

print(f"Ec = {energie_cinetique:.2f} J")
print(f"Ep = {energie_potentielle:.2f} J")
print(f"Em = {energie_mecanique:.2f} J")
````

### Vérifier
Remplace les intrants par chaque ligne du jeu d'essais et compare. Remarque l'écriture `0.5 * masse * vitesse ** 2` : grâce aux priorités, seule la vitesse est au carré. Écrire `(0.5 * masse * vitesse) ** 2` donnerait un résultat faux, sans aucun message d'erreur. **Python n'avertit jamais d'une erreur de logique** : seul ton jeu d'essais peut la détecter.

`G` est en majuscules par convention : c'est une **constante**, une valeur qu'on ne modifie pas.

## 7. Trois erreurs à reconnaître

| Erreur | Exemple | Cause |
|---|---|---|
| `TypeError` | `"Masse : " + 5` | opération impossible entre deux types (texte + nombre) |
| `ValueError` | `float("2,5")` | conversion impossible : la valeur n'a pas le bon format |
| `NameError` | `print(Masse)` | la variable n'existe pas (ici, à cause de la majuscule) |
